```python
import numpy as np
import matplotlib.pyplot as plt

def PowerFunction(upwelling_irradiance_mean, Area, test_hour_number, panel_efficiency, number_of_data):
    x = np.linspace(0,test_hour_number, number_of_data)
    number_of_oscilations = test_hour_number/.2 # .2 is the number of hour for 1 osc
    return x, panel_efficiency*Area*upwelling_irradiance_mean*np.sin(number_of_oscilations*np.pi*x)**2

x, sim_power = PowerFunction(210, .1245, 2, .2, 360*20*2)

plt.title(f'Simulated Upwelling Power V time')

plt.plot(x, sim_power)
plt.xlabel('time [Hours]')
plt.ylabel('Power [W]')
# plt.xticks(['1','2','3','4','5','6'])


```
